Comput the Paird t-test and ANOVa of full layered Hybrid model without GLS (Guided local search) agianest the two base models. This experiment was conducted using Ethiopian Traffic Sign dataset (ETSD)

```python
from scipy.stats import ttest_rel, f_oneway

# Example data for testing accuracy (from multiple runs)
MobileNet= [81.36, 81.23, 81.36, 81.95]
VGG16 = [96.36, 96.91, 96.18, 96.09]
Hybrid_model = [99.14, 99.32, 99.05, 98.73]



# Paired t-test (VGG16 vs Hybrid_model)
t_stat, p_value = ttest_rel(VGG16, Hybrid_model)
print("Paired t-test (VGG16 vs Hybrid_model):", t_stat, p_value)
# Paired t-test (Hybrid_model vs MobileNet)
t_stat, p_value = ttest_rel(MobileNet, Hybrid_model)
print("Paired t-test (MobileNet vs Hybrid_model):", t_stat, p_value)

# ANOVA for all three models
f_stat, p_value = f_oneway(Hybrid_model,VGG16, MobileNet)
print("ANOVA for all models:", f_stat, p_value)
```

# Mean standared deviation of Accuracy, F1-score, Precision and Recall

```python

# --- Basic statistical support for model comparison ---
import numpy as np
import pandas as pd
from scipy import stats

# ----------------------------
# 1️⃣  ENTER YOUR RESULTS HERE
# ----------------------------
# Example values: replace these with your own results for each seed
results = {
     "MobileNet": {
          "accuracy": [63.23, 71.23, 74.64, 75.05],
        "precision": [67.54, 73.54, 75.34, 56.94],
         "recall": [63.31, 71.23, 74.05, 56.05],
        "f1": [61.32, 69.91, 72.14, 53.06],
    },
    "VGG16": {
        "accuracy": [97.27, 96.38, 96.45, 97.23],
        "precision": [96.74, 97.90, 96.66, 97.66],
        "recall": [95.95, 97.27, 96.45, 97.23],
        "f1": [95.56, 97.29, 96.23, 97.02],
    },
    "Hybrid Model": {
        "accuracy": [99.14, 99.00, 98.91, 98.59],
        "precision": [99.09, 98.82, 98.85, 98.14],
        "recall": [99.09, 99.00, 98.81, 98.59],
        "f1": [98.83, 98.83, 98.85, 98.27],
    },
}

# ------------------------------------------------------
# 2️⃣  Function to compute mean, std, and 95% confidence interval
# ------------------------------------------------------
def summary_stats(values):
    n = len(values)
    mean = np.mean(values)
    std = np.std(values, ddof=1)
    t_value = stats.t.ppf(1 - 0.025, df=n - 1)  # 95% CI
    ci = t_value * std / np.sqrt(n)
    return mean, std, ci

# ------------------------------------------------------
# 3️⃣  Create summary table
# ------------------------------------------------------
rows = []
for model, metrics in results.items():
    for metric, vals in metrics.items():
        mean, std, ci = summary_stats(vals)
        rows.append([model, metric, mean, std, ci])

df = pd.DataFrame(
    rows,
    columns=["Model", "Metric", "Mean", "Std", "95% CI"]
)
df = df.round(4)
df = pd.DataFrame(rows, columns=["Model", "Metric", "Mean", "Std", "95% CI"])
print("=== Mean ± Std and 95% CI for each model ===")
display(df.pivot(index="Model", columns="Metric", values=["Mean", "Std", "95% CI"]))

# ------------------------------------------------------
# 4️⃣  Paired t-tests (Hybrid vs others) and ANOVA
# ------------------------------------------------------
metrics_to_test = ["accuracy", "precision", "recall", "f1"]
# metrics_to_test = ["precision", "recall", "f1"]

for metric in metrics_to_test:
    mob = results["MobileNet"][metric]
    vgg = results["VGG16"][metric]
    hyb = results["Hybrid Model"][metric]

    # paired t-tests
    t_mob, p_mob = stats.ttest_rel(mob, hyb)
    t_vgg, p_vgg = stats.ttest_rel(vgg, hyb)
    f_val, p_anova = stats.f_oneway(vgg, mob, hyb)

    print(f"\n=== {metric.upper()} ===")
    print(f"Paired t-test (VGG16 vs Hybrid):  t = {t_vgg:.4f},  p = {p_vgg:.6f}")
    print(f"Paired t-test (MobileNet vs Hybrid):  t = {t_mob:.4f},  p = {p_mob:.6f}")
    print(f"ANOVA (VGG16, MobileNet, Hybrid):  F = {f_val:.4f},  p = {p_anova:.6e}")

# ------------------------------------------------------
# 5️⃣  Interpretation helper
# ------------------------------------------------------
print("\nNote:")
print("- p < 0.05 → statistically significant difference.")
print("- 95% CI = mean ± CI margin.")
print("- Use mean ± std (or CI) values in your Tables 4–6.")

```
